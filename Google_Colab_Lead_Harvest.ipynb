### Step 3: Download Full 1.9M Shopify Master List from GitHub

Downloads shopify_master_list.csv.gz (51 MB compressed) from GitHub,
then decompresses it to the full 385 MB / 1.9M-row CSV.
On Colab's 1 Gbps connection: ~30s download + ~30s decompress.

```python
import os, gzip, time, subprocess

GZ_URL   = 'https://raw.githubusercontent.com/growthenginenowoslawski/shopify-master-list/main/shopify_master_list.csv.gz'
GZ_PATH  = '/content/data/shopify_master_list.csv.gz'
CSV_PATH = '/content/data/shopify_master_list.csv'

if os.path.exists(CSV_PATH) and os.path.getsize(CSV_PATH) > 100_000_000:
    print('[SKIP] CSV already decompressed: ' + str(os.path.getsize(CSV_PATH)) + ' bytes')
else:
    if not (os.path.exists(GZ_PATH) and os.path.getsize(GZ_PATH) > 40_000_000):
        print('[*] Downloading shopify_master_list.csv.gz (~51 MB) from GitHub...')
        t0 = time.time()
        subprocess.run(['wget', '-q', '--show-progress', '-O', GZ_PATH, GZ_URL], check=True)
        elapsed = time.time() - t0
        size_mb = os.path.getsize(GZ_PATH) / 1_048_576
        speed = size_mb / elapsed if elapsed > 0 else 0
        print('[OK] Downloaded ' + str(round(size_mb, 1)) + ' MB in ' + str(round(elapsed, 1)) + 's (' + str(round(speed, 1)) + ' MB/s)')
    else:
        print('[SKIP] GZ already cached: ' + str(os.path.getsize(GZ_PATH)) + ' bytes')

    print('[*] Decompressing to full 1.9M-row CSV...')
    t0 = time.time()
    rows = 0
    with gzip.open(GZ_PATH, 'rt', encoding='utf-8', errors='ignore') as fin, \
         open(CSV_PATH, 'w', encoding='utf-8') as fout:
        for line in fin:
            fout.write(line)
            rows += 1
            if rows % 500_000 == 0:
                print('  ... ' + str(rows) + ' rows')

    csv_mb = os.path.getsize(CSV_PATH) / 1_048_576
    print('[OK] ' + str(rows) + ' stores ready (' + str(round(csv_mb, 1)) + ' MB) in ' + str(round(time.time()-t0, 1)) + 's')

MASTER_CSV_PATH = CSV_PATH
print('[INFO] Master CSV: ' + MASTER_CSV_PATH)
```

### Step 4: Initialize Lead Output Files

```python
import csv, os, subprocess

CSV_HEADERS = [
    'First Name', 'Last Name', 'Email', 'Secondary_Email', 'Company', 'Website', 'Title',
    'Meta Ads Library URL', 'Niche', 'Country', 'Instagram', 'Facebook',
    'LinkedIn', 'TikTok', 'Twitter_X', 'MX_Status', 'DMARC_Status',
    'SPF_Status', 'Gravatar_Found', 'Deliverability_Score', 'Personalized Icebreaker'
]

niche_files = {
    'Jewelry':              '/content/data/verified_5k/jewelry_real_inboxes.csv',
    'Fashion & Apparel':    '/content/data/verified_5k/fashion_apparel_real_inboxes.csv',
    'Fashion Accessories':  '/content/data/verified_5k/fashion_accessories_real_inboxes.csv',
    'Health & Supplements': '/content/data/verified_5k/health_supplements_real_inboxes.csv',
    'Extended DTC':         '/content/data/verified_5k/extended_dtc_real_inboxes.csv',
    'Master':               '/content/data/verified_5k/instantly_master_ultra_verified.csv',
}

for name, fpath in niche_files.items():
    if not os.path.exists(fpath):
        with open(fpath, 'w', newline='', encoding='utf-8') as f:
            csv.writer(f).writerow(CSV_HEADERS)
        print('[+] Created: ' + os.path.basename(fpath))
    else:
        result = subprocess.run(['wc', '-l', fpath], capture_output=True, text=True)
        lines = int(result.stdout.strip().split()[0]) - 1 if result.returncode == 0 else '?'
        print('[=] ' + os.path.basename(fpath) + ': ' + str(lines) + ' existing leads')

print('[OK] Lead files ready.')
```

### Step 6: Audit Rejection Reasons (run anytime)
See exactly why each store did not qualify.

```python
import pandas as pd, os

raw_csv = '/content/data/raw_extracted/raw_extracted_all.csv'
if not os.path.exists(raw_csv):
    print('[!] No raw log yet. Run Step 5 first.')
else:
    df = pd.read_csv(raw_csv, low_memory=False)
    qual = df[df['Qualified'] == 'YES']
    rej  = df[df['Qualified'] == 'NO']
    print('Total visited   : ' + str(len(df)))
    print('Qualified (YES) : ' + str(len(qual)))
    print('Rejected  (NO)  : ' + str(len(rej)))
    print()
    print('Rejection breakdown:')
    print(rej['Rejection_Reason'].value_counts().head(20).to_string())
    print()
    no_mx = rej[rej['Rejection_Reason'].str.startswith('no_mx', na=False)]
    print('Stores with email found but no MX (recoverable): ' + str(len(no_mx)))
    if len(no_mx):
        print(no_mx[['Domain','Brand','Primary_Email','Rejection_Reason']].head(20).to_string(index=False))
```

### Step 7: Final Download (all leads + raw log in one zip)

```python
import os, zipfile, csv as csv_
from google.colab import files

out_zip = '/content/flinza_ALL_verified_leads.zip'
vdir    = '/content/data/verified_5k'
raw_csv = '/content/data/raw_extracted/raw_extracted_all.csv'

total_rows = 0
print('[*] Building final zip...')
with zipfile.ZipFile(out_zip, 'w', zipfile.ZIP_DEFLATED) as z:
    if os.path.isdir(vdir):
        for fname in sorted(os.listdir(vdir)):
            if fname.endswith('.csv'):
                fpath = os.path.join(vdir, fname)
                with open(fpath) as cf:
                    rows = sum(1 for _ in csv_.reader(cf)) - 1
                total_rows += rows
                z.write(fpath, arcname='verified/' + fname)
                print('  [+] ' + fname + ': ' + str(rows) + ' leads')
    if os.path.exists(raw_csv):
        z.write(raw_csv, arcname='raw_extracted_all.csv')
        print('  [+] raw_extracted_all.csv')

size_mb = os.path.getsize(out_zip) / 1_048_576
print('[OK] ' + str(total_rows) + ' verified leads -> ' + out_zip + ' (' + str(round(size_mb, 1)) + ' MB)')
files.download(out_zip)
```
